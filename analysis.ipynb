# Adidas Analysis Dashboard

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# import the dataset
data=pd.read_excel('Adidas.xlsx')
```

```python
data.isnull().sum()
```

```python
data[data.duplicated()]
```

### Total Sales

```python
data['OperatingProfit'].sum()
```

```python
data['TotalSales']-data['OperatingProfit']
```

```python
data['Retailer'].unique()
```

```python
data.groupby('Retailer')['TotalSales'].sum()
```

```python
# retailer-wise total sales bar chart
sales_by_retailer = data.groupby('Retailer')['TotalSales'].sum().sort_values(ascending=False)

plt.figure(figsize=(12,6))
sns.barplot(x=sales_by_retailer.index, y=sales_by_retailer.values, palette='viridis')
plt.title('Total Sales by Retailer')
plt.xlabel('Retailer')
plt.ylabel('Total Sales')
plt.xticks(rotation=45, ha='right')

# annotate bars
for i, v in enumerate(sales_by_retailer.values):
 plt.text(i, v, f'{v:,.0f}', ha='center', va='bottom', fontsize=8, rotation=90)

plt.tight_layout()
plt.show()
```
